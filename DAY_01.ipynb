```python
#Read a word and print whether it’s a palindrome (it reads the same backwards, like madam).
word = input()

if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")

```

```python
#Read n and print the sum of its digits. For example, 472 gives 13. (Hint: %10 gives the last digit, //10 removes it.)
n = int(input())

total = 0
while n > 0:
    digit = n % 10      # last digit
    total = total + digit
    n = n // 10         # remove the last digit

print(total)
```

```python
#count each letter with a dictionary
word = input()

count = {}
for ch in word:
    if ch in count:
        count[ch] += 1
    else:
        count[ch] = 1

for ch in count:
    print(ch, count[ch])


```
